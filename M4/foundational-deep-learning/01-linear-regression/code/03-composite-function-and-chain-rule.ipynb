```python
def fx(x):
  return 2*x - 1

def gf(fx):
  return (fx - 3) **2

def df_dx():
  return 2

def dg_df(fx):
  return 2*(fx - 3)
```

```python
lr = 0.1
x = 60

old_loss = gf(fx(x))
print(f"old_loss: {old_loss}")

# Công thức đúng theo Chain Rule: dg/dx = dg/df * df/dx
dg_df_val = dg_df(fx(x))
df_dx_val = df_dx()
dg_dx = dg_df_val * df_dx_val

print(f"dg_df: {dg_df_val}")
print(f"df_dx: {df_dx_val}")
print(f"dg_dx (Chain Rule chuẩn): {dg_dx}")

# Cập nhật x
x_new = x - lr * dg_dx
print(f"x mới: {x_new}")

new_loss = gf(fx(x_new))
print(f"new_loss mới: {new_loss}")
```
