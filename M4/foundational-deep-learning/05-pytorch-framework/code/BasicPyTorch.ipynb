## Broadcasting

````python
# create a tensor
tensor_1 = torch.tensor([[1,2],
                         [3,4]])

tensor_2 = torch.tensor([1])

# Addition between two tensors
tensor_3 = tensor_1 + tensor_2

print(f"Tensor 1: {tensor_1}")
print(f"Tensor 2: {tensor_2}")
print(f"tensor_3: {tensor_3}")
````

## Important functions

## Squared Difference

````python
# create two tensors
x = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
y = 5

# compute squared difference
squared_diff = (x - y) ** 2

print(f"x: {x}")
print(f"y: {y}")
print(f"Squared Difference: {squared_diff}")
````

## Mean Squared Error

````python
# create two tensors
x = torch.tensor([1, 2, 3, 4], dtype=torch.float32)
y = torch.tensor([5, 5, 5, 5], dtype=torch.float32)

# compute MSE
loss_function = torch.nn.MSELoss()
mse = loss_function(x, y)

print(f"x: {x}")
print(f"y: {y}")
print(f"MSE: {mse}")
````

# Gradient computation

````python
# create a tensor to compute gradients
x = torch.tensor(2, dtype = torch.float32, requires_grad=True)

# operation
y = x ** 2
z = 3*y + 2

# Run backward pass to compute gradients
z.backward()

# chain rule
# x -> y -> z
# dz/dy = 3
# dy/dx = 2x
# dz/dx = 3 * 2x = 6x = 6 * 2 = 12

print(f"z: {z}")
print(f"Gradient dz/dx (x.grad): {x.grad}")
````

## Variable: Represents a tensor whose values can be changed

````python
x = torch.tensor(2, dtype = torch.float32)

x = torch.autograd.Variable(x, requires_grad=True)
y = x ** 2
z = 3*y + 2

z.backward()

print(f"z: {z}")
print(f"Gradient dz/dx (x.grad): {x.grad}")
````

````python
x = torch.tensor([[1, 1],
                  [1, 1]], dtype = torch.float32)
x = x.requires_grad_(True)

y = x ** 2
z = 3*y + 2

z.backward()

print(f"x: {x}")
print(f"y: {y, y.dtype}")
print(f"z: {z}")
print(f"Gradient dz/dx (x.grad): {x.grad}")
````

**Lỗi RuntimeError**: `grad can be implicitly created only for scalar outputs` xảy ra vì hàm `.backward()` trong PyTorch chỉ có thể tự động tính đạo hàm mà không cần đối số truyền vào khi đầu ra (z) là một số vô hướng (scalar - tức là một tensor chỉ có 1 phần tử).

Trong code của bạn, `x` là một tensor kích thước 2x2, dẫn đến z cũng là một tensor kích thước 2x2 (không phải là một scalar).

Để khắc phục lỗi này, chúng ta có 2 cách phổ biến:

  - Cách 1 (Thường dùng trong học máy): Chuyển z thành một scalar bằng cách lấy tổng hoặc trung bình cộng (ví dụ: `z.sum().backward()`).
  - Cách 2: Truyền một tensor gradient cùng kích thước làm trọng số đạo hàm vào hàm backward, ví dụ: `z.backward(torch.ones_like(z))`.

````python
x = torch.tensor([[1, 1],
                  [1, 1]], dtype = torch.float32)
x = x.requires_grad_(True)

# y = x ** 2
y = x.sum()
z = 3*y + 2

z.backward()

print(f"x: {x}")
print(f"y: {y, y.dtype}")
print(f"z: {z}")
print(f"Gradient dz/dx (x.grad): {x.grad}")
````
